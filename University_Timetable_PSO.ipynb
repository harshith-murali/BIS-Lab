# University Timetable PSO

Particle swarm optimization for weekly course scheduling. Edit the problem and PSO parameters. Each particle encodes one timeslot index per course; fitness penalizes faculty clashes and daily imbalance.

Run the code cell to generate **`university-timetable-pso.png`** in the repository folder.

```python
# Editable inputs
COURSES=["CS101","CS102","CS103","EE101","EE102","MATH101"]
FACULTY=["Sharma","Patel","Sharma","Iyer","Iyer","Nair"]
TIMESLOTS=["Mon 9","Mon 11","Wed 9","Wed 11","Fri 9"]
PARTICLES, ITERATIONS, SEED=60,250,42
OUTPUT="university-timetable-pso.png"
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(SEED)
days=[s.split()[0] for s in TIMESLOTS]; ds=set(days); target=len(COURSES)/len(ds)
def cost(x):
 c=sum(x[i]==x[j] and FACULTY[i]==FACULTY[j] for i in range(len(x)) for j in range(i)); loads=[sum(days[s]==d for s in x) for d in ds]; return 1000*c+sum((n-target)**2 for n in loads)
n=len(COURSES); X=rng.uniform(0,len(TIMESLOTS)-1,(PARTICLES,n)); V=rng.uniform(-1,1,X.shape); P=X.copy(); pc=np.array([cost(np.rint(x).astype(int)) for x in X]); g=P[pc.argmin()].copy(); gc=pc.min(); hist=[gc]
for t in range(ITERATIONS):
 w=.9-.5*t/max(1,ITERATIONS-1); V=np.clip(w*V+1.6*rng.random(X.shape)*(P-X)+1.6*rng.random(X.shape)*(g-X),-2,2); X=np.clip(X+V,0,len(TIMESLOTS)-1); sc=np.array([cost(np.rint(x).astype(int)) for x in X]); take=sc<pc; P[take]=X[take]; pc[take]=sc[take]; j=pc.argmin()
 if pc[j]<gc: g=P[j].copy(); gc=pc[j]
 hist.append(gc)
 if gc==0: break
best=np.rint(g).astype(int); print("Best penalty:",cost(best))
for c,f,s in zip(COURSES,FACULTY,best): print(f"{c:10} {f:10} {TIMESLOTS[s]}")
fig,(a,b)=plt.subplots(2,1,figsize=(12,9)); a.scatter(best,range(n),s=650,c=range(n),cmap="Set2")
for i,(c,f) in enumerate(zip(COURSES,FACULTY)): a.text(best[i],i,f"{c}\n{f}",ha="center",va="center",fontsize=8)
a.set(xticks=range(len(TIMESLOTS)),xticklabels=TIMESLOTS,yticks=[],title="Best timetable"); a.set_ylim(len(COURSES)-.35,-.65); a.grid(axis="x",alpha=.2); b.plot(hist); b.set(title="Global best",xlabel="Iteration",ylabel="Penalty"); b.grid(alpha=.2); fig.tight_layout(); fig.savefig(OUTPUT,dpi=160); plt.show(); print("Saved",OUTPUT)
```
